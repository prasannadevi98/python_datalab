```python

import numpy as np
```

```python
array_1=np.array([10,20,30])
print(array_1)
print(array_1.shape)
print(array_1.size)
print(array_1.ndim)
print(array_1.dtype)

```

```python
array_2=np.array([[10,20,330],[40,50,60,]])
print (array_2)
print(array_2.shape)
print(array_2.size)
print(array_2.ndim)


```

```python
array_3= np.array([[[10,20,30],[40,50,60],[70,80,90]]])
print(array_3)
print(array_3.shape)
print(array_3.size)
print(array_3.ndim)
```

### ** Array filling*

```python
a=np.full((2,2),2)
print(a)
```

```python
a=np.full((2,2,2),[3])
print(a)
```

```python
a=np.zeros((3,3))
print (a)
```

```python
a=np.zeros((3,3),dtype="int")
print(a)
           
```

### **ARANGE*

```python
for each in range (-10,11,2):
    print(each,end=" ")
```

```python
x=np.arange(-10,11,2)
print(x)
```

### ** linspace*
