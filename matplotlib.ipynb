### ** matplotlib -linestyle*

```python
l1=np.array([1,2,3,4])
l2=np.array([5,6,7,8])
plt.plot(l1,l2,c="g",linestyle="dashed")
plt.scatter(l1,l2,c="g",marker="*",s=100)
plt.show();
```

### * bar chart*

```python
year=["1998","1999","2005","2010"]
medicine=["calpal","paracetomal","citizin","dolo"]
c= "green","yellow","red","blue"
plt.bar(year,medicine,color=c,align="edge")
plt.show()
```

### **pie chart*

```python
medicine=["calpal","paracetomal","citizin","dolo"]
price=[2,5,3,6]
exp=[0.05,0.05,0.05,0.05]
c=["green","yellow","blue","red"]
plt.pie(price,labels= medicine,explode=exp,colors=c)
plt.show();
```

```python
years=["1998","1999","2005","2010"]
tablet_price=[2,5,3,6]
plt.title("price range acroosing year")
plt.plot(year ,tablet_price)
plt.scatter(years,tablet_price,color="g",marker="*")
plt.show();
```
